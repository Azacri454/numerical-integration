```python
import math
def f(x):
    return math.exp(-x ** 2)

def trapeze_composite(f, a, b, n):
    h = (b - a) / n
    S = f(a) + f(b)
    for i in range(1, n):         
        x = a + i * h
        S = S + 2 * f(x)
    return (h / 2) * S
a, b = 0, 1
for n in [4, 6]:
    I = trapeze_composite(f, a, b, n)
    print("n =", n, " I =", I)
```

```python
import math

def f(x):
    return math.exp(-x ** 2)

def simpson_composite(f, a, b, n):
    if n % 2 != 0:
        raise ValueError("n doit etre pair")
    h = (b - a) / n
    I = f(a)                      
    for i in range(1, n):          
        x = a + i * h
        if i % 2 == 1:            
            I = I + 4 * f(x)
        else:                      
            I = I + 2 * f(x)
    I = I + f(b)                  
    return I * h / 3
a, b = 0, 1
for n in [4, 6]:
    I = simpson_composite(f, a, b, n)
    print("n =", n, " I =", I)
```
